```python
from pathlib import Path 
import pandas as pd 
import numpy as np 


project_root = Path.cwd().parent

master_path = (
    project_root
    / "data"
    / "processed"
    / "five_patient_combined_features.csv"
)

master_df = pd.read_csv(master_path)
print(master_df.shape)
master_df.head()
```

For ML dataset:
X = 9 EEG feature columns (input)
Y = seizure / non-seizure label (output)


```python
feature_columns = [
    "mean_uv",
    "std_uv",
    "rms_uv",
    "ptp_uv",
    "delta_power",
    "theta_power",
    "alpha_power",
    "beta_power",
    "gamma_power"
]

X = master_df[feature_columns]
y = master_df["label"]

print("X shape:", X.shape)
print("y shape:", y.shape)

print("\nLabels:")
print(y.value_counts())
```

```python
train_mask = master_df["patient_id"].isin(["chb01", "chb02", "chb03"])
validation_mask = master_df["patient_id"] == "chb05"
test_mask = master_df["patient_id"] == "chb08"

x_train = master_df.loc[train_mask, feature_columns]
y_train = master_df.loc[train_mask, "label"]

x_validation = master_df.loc[validation_mask, feature_columns]
y_validation = master_df.loc[validation_mask, "label"]

x_test = master_df.loc[test_mask, feature_columns]
y_test = master_df.loc[test_mask, "label"]

print("Training:", x_train.shape, y_train.shape)
print("Validation:", x_validation.shape, y_validation.shape)
print("Test:", x_test.shape, y_test.shape)

print("\nTraining labels:")
print(y_train.value_counts())

print("\nValidation labels:")
print(y_validation.value_counts())

print("\nTest labels:")
print(y_test.value_counts())
```

Next I have to fix the feature scaling, since some values are computed at a much larger value than others.

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

#fit training data
x_train_scaled = scaler.fit_transform(x_train)

#transform validation data
x_validation_scaled = scaler.transform(x_validation)

#transform testing data
x_test_scaled = scaler.transform(x_test)
```

```python
print("Training feature means:")
print(x_train_scaled.mean(axis=0))

print("\nTraining feature standard deviations:")
print(x_train_scaled.std(axis=0))
```
